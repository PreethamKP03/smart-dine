```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
DATA_PATH = r"C:\Users\MEGHANA\OneDrive\Desktop\SmartDine\data\processed\group_restaurant_features.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
print()
print("Columns:")
print(df.columns.tolist())
```

```python
print("Suitability score statistics:")
print(
    df["suitability_score"].describe()
)
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["suitability_score"],
    bins=30,
    kde=True
)

plt.title("Distribution of Group-Restaurant Suitability Scores")
plt.xlabel("Suitability Score")
plt.ylabel("Number of Group-Restaurant Pairs")

plt.show()
```

```python
feature_columns = [
    "group_size",
    "cuisine_match",
    "budget_match",
    "dietary_compatibility",
    "restaurant_type_match",
    "location_match",
    "booking_match",
    "online_order_match",
    "group_agreement",
    "rating_score",
    "popularity_score"
]

target_correlations = (
    df[feature_columns + ["suitability_score"]]
    .corr()["suitability_score"]
    .sort_values(ascending=False)
)

print(
    "Correlation with suitability score:"
)

print(target_correlations)
```

```python
print("Highest suitability scores:")

print(
    df[
        [
            "group_id",
            "restaurant_id",
            "cuisine_match",
            "budget_match",
            "dietary_compatibility",
            "restaurant_type_match",
            "location_match",
            "group_agreement",
            "rating_score",
            "popularity_score",
            "suitability_score"
        ]
    ]
    .sort_values(
        "suitability_score",
        ascending=False
    )
    .head(10)
    .to_string(index=False)
)
```
